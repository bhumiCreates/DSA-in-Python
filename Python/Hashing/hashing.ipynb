#### A common rule of thumb:

#### - numbers, such as 0–10⁵: A frequency array is often suitable.
#### - Numbers up to 10⁶: A frequency array may be suitable, depending on memory limits.
#### - Very large numbers, such as 10⁹, or negative numbers: Prefer a dictionary.
#### - Unknown input constraints: Check the problem statement before choosing the data structure.

```python
n = int(input("Enter the number "))
arr = list(map(int, input("Enter the digits of array ").split()))

hash_arr = [0] * 100001

for num in arr:
    hash_arr[num] += 1

q = int(input("Enter the number"))

while q > 0:
    num = int(input("enter the number to fetch"))
    print(hash_arr[num])
    q -= 1
```

#### A dictionary stores entries only for numbers that appear, rather than allocating a position for every possible number.

```python
arr = [1000000000, 5, 10, 12, 12, 1000000000]

hash_map = {}

for num in arr:
    hash_map[num] = hash_map.get(num, 0) + 1

print(hash_map)
```

#### Character Hashing

#### Note: Python doesn't allow character subtraction like s[i] - 'a'.We need to use ord() to convert characters into ASCII values.

#### For this code it only takes 26 letters which are lower case letters but if it is not specified whether the letters are in lower case or upper case we must take count of letters as **256**

```python
s=str(input("Enter the string : "))
hash_arr = 26 * [0]

for i in range(len(s)):
    hash_arr[ord(s[i]) - ord('a')] += 1
    
q=int(input("Enter the number of queries: "))

while q>0:
    c=str(input("Enter the character you want to search: "))
    print(hash_arr[ord(c)-ord('a')])
    q-=1
```
