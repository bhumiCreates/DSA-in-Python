#### Leetcode Ques. 1365 : Numbers smaller than current number 
#### Link-> https://leetcode.com/problems/how-many-numbers-are-smaller-than-the-current-number/

```python
class Solution(object):
    def smallerNumbersThanCurrent(self, nums):
        arr=[]
        for i in nums:
            count=0
            for j in nums:
                if j<i:
                    count+=1
            arr.append(count)

        return arr
```

#### Leetcode Ques. 412 FizzBuzz

```python
class Solution(object):
    def fizzBuzz(self, n):
        answer=[]

        for i in range(1,n+1):
            if i%3==0 and i%5==0:
                i="FizzBuzz"
                answer.append(i)
            elif i%5==0:
                i="Buzz"
                answer.append(i)
            elif i%3==0 :
                i="Fizz"
                answer.append(i)
            else:
                answer.append(str(i))

        return answer  
            
        print(answer)
```

#### Leetcode ques. 1523 -> Count Odd Numbers in an Interval Range

```python
class Solution(object):
    def countOdds(self, low, high):
        return (high+1)//2-(low//2)
```

#### Leetcode ques. 2520 Count the Digits That Divide a Number

```python
class Solution(object):
    def countDigits(self, num):
        count = 0
        temp = num

        while temp > 0:
            digit = temp % 10

            if num % digit == 0:
                count += 1

            temp //= 10

        return count
```

#### Leetcode ques. 1281 Subtract the Product and Sum of Digits of an Integer

```python
class Solution(object):
    def subtractProductAndSum(self, n):
        temp=n
        sum=0
        product=1

        while temp>0:
            digit=temp%10
            product*=digit
            sum+=digit
            temp//=10
            result=product-sum
            
        return result
        
```
