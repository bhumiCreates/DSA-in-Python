#### When a function calls itself until a specific condition is met its called recursion. Loops are more efficient to use but can't be implemented in every situation.

#### Stack overflow/segmentation fault : when a fucntion keeps waiting for final execution but never gets to it becuase of an infinite condition

#### Base condition: the condition use to stop the code from infinite execution

```python
def fun1(i):
    if i==5:
        return
    
    print(i-1,end=" ")
    fun1(i+1)

fun1(0)
```

```python
def printName(i,n):
    if i>=n:
        return
    
    print("Bhumi", end="")
    print()
    printName(i+1,n)

    
printName(0,5)
    
```

```python
def printN(i,n):
    if i>n:
        return
    
    print(i,end="")
    print()
    printN(i+1,n)
    
printN(1,5)
```

```python
def reverseList(i,n):
    if i>=n:
        return
    
    print(n-i,end="")
    print()
    reverseList(i+1,n)
    
reverseList(0,5)
```

#### Bakctracking : Backtracking is a problem-solving technique where you try a possible solution, and if it doesn't work, you undo your choice and try another one.

#### (Print after upgrading the function)

```python
def printNum(i,n):
    if i<=0:
        return
    printNum(i-1,n)
    print(i,end="")  
    print()

printNum(5,5)
```

```python
def sum_n(n):
    if n == 0:
        return 0
    return n + sum_n(n - 1)


print(sum_n(6))
```

#### Recursion using two pointers

```python
def rev_arr(arr,i,j):
    if i>=j:
        return
    arr[i], arr[j] = arr[j], arr[i]
    
    rev_arr(arr,i+1,j-1)
    
arr=[12,33,43,21,34,77]

rev_arr(arr,0,len(arr)-1)
print(arr)

```

```python
# Single variable
arr=[1,2,3,4,5]
n=len(arr)
def reverse_array(i):
    if i>n/2:
        return
    
    arr[i],arr[n-i-1] = arr[n-i-1],arr[i]
    reverse_array(i+1)
    
reverse_array(0)
print(arr)
    
```

#### Multiple Recursion calls

```python
def fibonacci_num(n):
    if n<=1:
        return n
    return fibonacci_num(n-1)+fibonacci_num(n-2)

print(fibonacci_num(8))

# Time complexity : O(2^n) - exponential
```
